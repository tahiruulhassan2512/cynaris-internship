```python
import sqlite3
import pandas as pd

# Create in-memory sales database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create sales table
cursor.execute('''
CREATE TABLE sales (
    id INTEGER PRIMARY KEY,
    product TEXT,
    region TEXT,
    amount DECIMAL(10,2),
    sale_date DATE,
    quantity INTEGER,
    discount DECIMAL(4,2)
)
''')

# Insert sample sales data
cursor.executemany(
    "INSERT INTO sales VALUES (?, ?, ?, ?, ?, ?, ?)",
    [
        (1, 'Laptop', 'North', 85000, '2026-01-15', 2, 0.05),
        (2, 'Phone', 'South', 45000, '2026-02-20', 5, 0.10),
        (3, 'Tablet', 'East', 32000, '2026-01-28', 3, 0.00),
        (4, 'Laptop', 'West', 85000, '2026-03-10', 1, 0.15),
        (5, 'Phone', 'North', 45000, '2026-02-05', 4, 0.05),
        (6, 'Headphones', 'South', 8000, '2026-03-15', 10, 0.00),
        (7, 'Tablet', 'North', 32000, '2026-04-01', 2, 0.00),
        (8, 'Laptop', 'East', 85000, '2026-01-20', 3, 0.10),
        (9, 'Phone', 'West', 45000, '2026-04-18', 6, 0.00),
        (10, 'Headphones', 'East', 8000, '2026-02-28', 8, 0.00),
        (11, 'Tablet', 'South', 32000, '2026-03-22', 1, 0.05),
        (12, 'Laptop', 'South', 85000, '2026-05-10', 2, None)
    ]
)

conn.commit()

# Query 1 - WHERE
result1 = pd.read_sql(
    "SELECT * FROM sales WHERE region = 'North'",
    conn
)
print("Query 1:")
print(result1)

# Query 2 - WHERE with AND
result2 = pd.read_sql(
    "SELECT * FROM sales WHERE region = 'North' AND amount > 50000",
    conn
)
print("\nQuery 2:")
print(result2)

# Query 3 - WHERE with OR
result3 = pd.read_sql(
    "SELECT * FROM sales WHERE region = 'North' OR region = 'South'",
    conn
)
print("\nQuery 3:")
print(result3)

# Query 4 - WHERE with NOT
result4 = pd.read_sql(
    "SELECT * FROM sales WHERE NOT region = 'North'",
    conn
)
print("\nQuery 4:")
print(result4)

# Query 5 - LIKE
result5 = pd.read_sql(
    "SELECT * FROM sales WHERE product LIKE 'L%'",
    conn
)
print("\nQuery 5:")
print(result5)

# Query 6 - IN
result6 = pd.read_sql(
    "SELECT * FROM sales WHERE region IN ('North', 'East')",
    conn
)
print("\nQuery 6:")
print(result6)

# Query 7 - BETWEEN
result7 = pd.read_sql(
    "SELECT * FROM sales WHERE amount BETWEEN 30000 AND 70000",
    conn
)
print("\nQuery 7:")
print(result7)

# Query 8 - IS NULL
result8 = pd.read_sql(
    "SELECT * FROM sales WHERE discount IS NULL",
    conn
)
print("\nQuery 8:")
print(result8)

# Query 9 - Comparison operator
result9 = pd.read_sql(
    "SELECT * FROM sales WHERE quantity >= 5",
    conn
)
print("\nQuery 9:")
print(result9)

# Query 10 - Combined conditions
result10 = pd.read_sql(
    "SELECT * FROM sales WHERE amount > 50000 AND discount < 0.10",
    conn
)
print("\nQuery 10:")
print(result10)

# Export one result set to CSV
result8.to_csv("sales_null_discount.csv", index=False)

print("\nCSV export completed.")
```
